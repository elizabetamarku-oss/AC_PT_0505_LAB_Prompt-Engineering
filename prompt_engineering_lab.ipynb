Part 1: Starting Simple - Create Initial Prompts
Objective: Create basic zero-shot prompts for three different task types and establish a baseline.

````python
import os
from dotenv import load_dotenv
from openai import OpenAI

# ---------------------------------------------------------
# 1. LOAD API KEY FROM .env (kept hidden)
# ---------------------------------------------------------
load_dotenv()  # loads .env into environment variables

api_key = os.getenv("OPENAI_API_KEY")
if not api_key:
    raise ValueError("❌ OPENAI_API_KEY not found. Check your .env file.")

client = OpenAI(api_key=api_key)

# ---------------------------------------------------------
# 2. HELPER FUNCTION TO CALL OPENAI
# ---------------------------------------------------------
def call_openai(prompt: str, model: str = "gpt-4o-mini") -> str:
    """
    Sends a prompt to OpenAI and returns the text output.
    """
    response = client.responses.create(
        model=model,
        input=prompt
    )
    return response.output_text.strip()


# ---------------------------------------------------------
# 3. RUN PROMPT MULTIPLE TIMES
# ---------------------------------------------------------
def run_prompt_n_times(prompt: str, n: int, label: str = ""):
    print("\n" + "=" * 60)
    print(f"{label} — running {n} times")
    print("=" * 60)

    outputs = []

    for i in range(1, n + 1):
        result = call_openai(prompt)
        outputs.append(result)
        print(f"\n--- Run {i} ---")
        print(result)

    unique = len(set(outputs))
    consistency = 100 * (1 - unique / n)

    print("\nSummary:")
    print(f"- Total runs: {n}")
    print(f"- Unique responses: {unique}")
    print(f"- Consistency: {consistency:.1f}%")

    return outputs
````

````python
# ---------------------------------------------------------
# Part 1: Create Initial Prompts
# ---------------------------------------------------------

#Initial simple prompt for sentiment analysis, product description, and data extraction. We will run each of these prompts multiple times to see how consistent the outputs are.
sentiment_prompt_v1 = """
Classify this customer message: "I love this product! It's exactly what I needed."
"""
product_prompt_v1 = """
Describe the product mentioned in this customer message:
"I love this product! It's exactly what I needed."
"""
extraction_prompt_v1 = """
Extract information from this customer feedback:
"I ordered item #12345 on March 15th. The delivery was fast but the packaging was damaged."
"""
 
#Test it once
print("Sentiment (single run):")
print(call_openai(sentiment_prompt_v1))

print("\nProduct Description (single run):")
print(call_openai(product_prompt_v1))

print("\nData Extraction (single run):")
print(call_openai(extraction_prompt_v1))
````

````python
# ---------------------------------------------------------
# Part 2: Diagnosing Failures - Systematic Testing (5 times each)
# ---------------------------------------------------------

run_prompt_n_times(sentiment_prompt_v1, 5, "Sentiment v1 (5 runs)")
run_prompt_n_times(product_prompt_v1, 5, "Product v1 (5 runs)")
run_prompt_n_times(extraction_prompt_v1, 5, "Extraction v1 (5 runs)")


````

````python
# ---------------------------------------------------------
# Part 2.1: Diagnosing Failures - Systematic Testing (10 times each)
# ---------------------------------------------------------

run_prompt_n_times(sentiment_prompt_v1, 10, "Sentiment v1 (10 runs)")
run_prompt_n_times(product_prompt_v1, 10, "Product v1 (10 runs)")
run_prompt_n_times(extraction_prompt_v1, 10, "Extraction v1 (10 runs)")

````

````python
# ---------------------------------------------------------
# Part 2.2: Diagnosing Failures - Systematic Testing (15 times each)
# ---------------------------------------------------------

run_prompt_n_times(sentiment_prompt_v1, 15, "Sentiment v1 (15 runs)")
run_prompt_n_times(product_prompt_v1, 15, "Product v1 (15 runs)")
run_prompt_n_times(extraction_prompt_v1, 15, "Extraction v1 (15 runs)")
````

Summary of Consistency Results

Across multiple runs, consistency ranged only between 0% and 20%.
The more times we ran the same zero‑shot prompt, the lower the consistency became.

Out of all runs, we saw almost 100% unique responses, meaning:
xThe model almost never repeated the same answer
xThe prompt is highly unstable
xZero‑shot prompts behave unpredictably
xThis level of variation is expected at the start of the lab
xWe are still far from the target of 80%+ consistency

This confirms why zero‑shot prompts are unreliable and why the next steps (few‑shot, structure, constraints, CoT) are essential for improving stability.

````python
# ---------------------------------------------------------
# Part 3: Iteration 1 - Improved Sentiment Analysis Prompt (v2)
# ---------------------------------------------------------

sentiment_prompt_v2 = """
You are a sentiment classifier.

Task:
Classify the sentiment of the following customer message.

Rules:
- Respond with ONE WORD ONLY.
- Allowed outputs: Positive, Negative, Neutral.
- Do NOT add explanations.

Message:
"I love this product! It's exactly what I needed."

Output:
"""

#Test it once
print("Sentiment (single run):")
print(call_openai(sentiment_prompt_v2))
````

````python
# ---------------------------------------------------------
# Part 3: Iteration 1 - Improved Product Description Prompt (v2)
# ---------------------------------------------------------

product_prompt_v2 = """
Write a product description following the rules below.

Product:
Wireless Bluetooth Mouse
Price: $29.99
Key Features: Ergonomic design for comfortable use
Technical Specifications:
- 2.4GHz wireless connectivity (10m range)
- Adjustable DPI: 800 / 1200 / 1600

Output Format (follow EXACTLY):
Title: <short product title>
Price: <price>
Text: 2 sentences describing features, specs, and benefits>

Rules:
- Max 40 words in the text section.
- Style: friendly, clear, concise.
- Do NOT include technical jargon.
- Do NOT add any text before or after the Title/Text.
- Do NOT include metadata, comments, or explanations.
- Do NOT use code fences.

Output:
"""
#Test it once
print("\nProduct Description (single run):")
print(call_openai(product_prompt_v2))
````

````python
# ---------------------------------------------------------
# Part 3: Iteration 1 - Improved Data Extraction Prompt (v2)
# ---------------------------------------------------------

extraction_prompt_v2 = """
Extract structured data from the customer feedback below.

Feedback:
"I ordered item #12345 on March 15th. The delivery was fast but the packaging was damaged."

Return ONLY valid JSON with these fields:
{
  "item_number": "",
  "order_date": "",
  "delivery_experience": "",
  "packaging_condition": ""
}

Rules:
- Return ONLY the JSON object.
- Do NOT add text before or after the JSON.
- Do NOT use code fences (no ```json or ```).
- Do NOT add explanations or metadata.
- Do NOT change field names.
- If information is missing, return an empty string.

Output:
"""

#Test it once
print("\nData Extraction (single run):")
print(call_openai(extraction_prompt_v2))
````

````python
# ---------------------------------------------------------
# Part 3: Diagnosing Failures V2 - Systematic Testing (15 times each)
# ---------------------------------------------------------

run_prompt_n_times(sentiment_prompt_v2, 15, "Sentiment v2 (15 runs)")
run_prompt_n_times(product_prompt_v2, 15, "Product v2 (15 runs)")
run_prompt_n_times(extraction_prompt_v2, 15, "Extraction v2 (15 runs)")
````

````python
#============================================================
#Part 4: Iteration 3 - Adding Structure & Constraints (Sentiment Analysis v3 — Few‑Shot)
#============================================================

sentiment_prompt_v3 = """
You are a sentiment classifier.

Format:
- Respond with ONE WORD ONLY.
- Allowed outputs: Positive, Negative, Neutral.

Examples:
Message: "This is the best purchase I've ever made."
Output: Positive

Message: "I'm very disappointed, it stopped working after one day."
Output: Negative

Message: "It's okay, nothing special but not terrible."
Output: Neutral

Now classify the sentiment of this message:
"I love this product! It's exactly what I needed."

Output:
"""
print(call_openai(sentiment_prompt_v3))
run_prompt_n_times(sentiment_prompt_v3, 15, "Sentiment v3 (15 runs)")
````

````python
#============================================================
#Part 4: Iteration 3 - Adding Structure & Constraints (Data Extraction v3 — Chain‑of‑Thought + JSON)
#============================================================

data_extraction_prompt_v3 = """
You are an information extraction assistant.

Task:
Extract structured data from the customer feedback.

Feedback:
"I ordered item #12345 on March 15th. The delivery was fast but the packaging was damaged."

First, think step-by-step (in your head) about:
- What is the item number?
- What is the order date?
- How was the delivery experience described?
- How was the packaging described?

Then output ONLY the final JSON object.

Return ONLY valid JSON with these fields:
{
  "item_number": "",
  "order_date": "",
  "delivery_experience": "",
  "packaging_condition": ""
}

Rules:
- Do NOT show your reasoning.
- Do NOT add any text before or after the JSON.
- Do NOT use code fences.
- Do NOT change field names.
- If information is missing, return an empty string.

Output:
"""

print(call_openai(data_extraction_prompt_v3))
run_prompt_n_times(data_extraction_prompt_v3, 15, "Data Extraction v3 (15 runs)")
````

````python
#============================================================
#Part 4: Iteration 3 - Adding Structure & Constraints (Product Description v3 — Chain‑of‑Thought + JSON)
#============================================================

product_prompt_v3 = """
You are a product copywriter.

Output Format (follow EXACTLY):
Title: <short product title>
Price: <product price>
Text: <2–3 sentences, max 30 words>

Examples:
Product: Bluetooth Speaker, $49.99
Title: Compact Bluetooth Speaker
Text: This portable speaker delivers clear, room-filling sound with easy Bluetooth pairing. Ideal for home or travel, it offers long battery life and reliable performance for just $49.99.

Product: USB-C Charger, $19.99
Title: Fast USB-C Charger
Text: This compact charger powers your devices quickly and safely. Perfect for phones and tablets, it’s designed for everyday use and great value at $19.99.

Now write a description for this product:

Product:
Wireless Bluetooth Mouse
Price: $29.99
Key Features: Ergonomic design for comfortable use
Technical Specifications:
- 2.4GHz wireless connectivity (10m range)
- Adjustable DPI: 800 / 1200 / 1600

Rules:
- Max 40 words in the Text section.
- Style: friendly, clear, concise.
- Include features, specs, and price.
- Do NOT add any text before or after Title/Text.
- Do NOT include metadata, comments, or explanations.
- Do NOT use code fences.

Output:
"""

print(call_openai(product_prompt_v3))
run_prompt_n_times(product_prompt_v3, 15, "Product v3 (15 runs)")
````

FINAL SUMMARY V3

After adapting all prompts with structure, constraints, and advanced techniques (few‑shot examples and Chain‑of‑Thought), consistency improved dramatically across most tasks:

✔ Sentiment Analysis — Consistency > 90%
Adding few‑shot examples and enforcing a strict one‑word output format produced highly stable results.
Across 15 runs, the model returned the same sentiment classification in nearly every attempt.

✔ Data Extraction — Consistency > 90%
Introducing hidden Chain‑of‑Thought reasoning and forcing a strict JSON‑only output significantly improved stability.
The model consistently extracted the same fields with minimal variation, even across multiple runs.

✘ Product Description — Consistency ~0%
Despite adding structure, examples, and word limits, the product description task remained highly variable.
Because the output is free‑form natural language, the model naturally introduces stylistic variation, making it difficult to achieve high consistency.
This is expected: creative text generation is inherently less stable than classification or structured extraction.
